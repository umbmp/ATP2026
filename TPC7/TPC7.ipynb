### Simplificação de frações

- Como simplificamos frações?
- Algoritmo de Euclides vs `math.gcd`.

````python
import math
math.gcd(10,5)
````

````python
#maximo divisor comum
# / float    // divisao inteira
def simplificarFracao(f):
    mdc = math.gcd(f[0], f[1])
    return (f[0]//mdc, f[1]// mdc)

print(simplificarFracao((55, 80)))





def mdc(a, b):
    while a % b != 0 :
        a, b       =      b, a % b
    return b

def simplificarFracao2(f):
    return (f[0]// mdc(f[0], f[1]), f[1]// mdc(f[0], f[1]))
print(simplificarFracao2((55, 80)))
````
